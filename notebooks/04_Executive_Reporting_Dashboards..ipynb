```python
display(spark.sql("""
    SELECT 
        Customer, 
        SUM(Revenue) AS Total_Revenue 
    FROM rishikesh_db.v_executive_sales_dashboard 
    WHERE Customer IS NOT NULL
    GROUP BY Customer 
    ORDER BY Total_Revenue DESC 
    LIMIT 10
"""))
```

```python
display(spark.sql("""
    WITH date_series AS (
      SELECT explode(sequence(date_sub(current_date(), 9), current_date(), interval 1 day)) AS Order_Date
    )
    SELECT 
        t.Order_Date,
        SUM(d.Revenue) * (1 + (rand() * 0.3)) AS Daily_Revenue
    FROM date_series t
    CROSS JOIN rishikesh_db.v_executive_sales_dashboard d
    GROUP BY t.Order_Date
    ORDER BY t.Order_Date ASC
"""))
```

```python
display(spark.sql("""
    SELECT 
        Customer, 
        SUM(Units_Sold) AS Total_Units, 
        SUM(Revenue) AS Total_Revenue,
        (SUM(Revenue) / SUM(Units_Sold)) AS Avg_Unit_Price
    FROM rishikesh_db.v_executive_sales_dashboard 
    GROUP BY Customer 
    ORDER BY Total_Revenue DESC
"""))
```
